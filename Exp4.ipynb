```python
# ============================================================
# PRACTICAL: DATA PREPROCESSING
# Dataset: weatherAUS.csv
# ============================================================

# 1. Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler, LabelEncoder
from google.colab import files

# ============================================================
# 2. Load Dataset
# ============================================================

uploaded = files.upload()

df = pd.read_csv('weatherAUS.csv')

print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

display(df.head())


# ============================================================
# 3. Study Skewness using Histogram and KDE
# ============================================================

# Select numerical columns
numeric_cols = df.select_dtypes(include=np.number).columns

# Calculate number of rows for subplots
n = len(numeric_cols)
rows = (n + 3) // 4

fig, axes = plt.subplots(rows, 4, figsize=(18, 4 * rows))
axes = axes.flatten()

for i, col in enumerate(numeric_cols):
    sns.histplot(
        df[col].dropna(),
        bins=30,
        kde=True,
        ax=axes[i]
    )

    axes[i].set_title(
        f'{col} | Skewness = {df[col].skew():.2f}'
    )
    axes[i].set_xlabel(col)
    axes[i].set_ylabel("Frequency")

# Remove unused plots
for i in range(n, len(axes)):
    fig.delaxes(axes[i])

plt.tight_layout()
plt.show()

# Display skewness values
print("Skewness of Numerical Attributes:")
display(
    df[numeric_cols]
    .skew()
    .sort_values(ascending=False)
)


# ============================================================
# 4. Select Numerical Features for Scaling
# ============================================================

features = [
    'MinTemp',
    'MaxTemp',
    'Rainfall',
    'Humidity9am',
    'Pressure9am'
]

# Remove missing values
data = df[features].dropna().copy()

print("Data used for Scaling:")
print("Shape:", data.shape)
display(data.head())


# ============================================================
# 5. Min-Max Scaling Manually using Pandas
# ============================================================

manual_minmax = (
    data - data.min()
) / (
    data.max() - data.min()
)

print("Manual Min-Max Scaling:")
display(manual_minmax.head())


# ============================================================
# 6. Min-Max Scaling using Scikit-Learn
# ============================================================

scaler = MinMaxScaler()

sklearn_minmax = pd.DataFrame(
    scaler.fit_transform(data),
    columns=features,
    index=data.index
)

print("Scikit-Learn MinMaxScaler:")
display(sklearn_minmax.head())


# ============================================================
# 7. Compare Manual and Scikit-Learn Scaling
# ============================================================

print("Comparison: Manual vs Scikit-Learn")

comparison = pd.DataFrame({
    'Manual_MinMax_MinTemp': manual_minmax['MinTemp'],
    'Sklearn_MinMax_MinTemp': sklearn_minmax['MinTemp'],
    'Manual_MinMax_MaxTemp': manual_minmax['MaxTemp'],
    'Sklearn_MinMax_MaxTemp': sklearn_minmax['MaxTemp']
})

display(comparison.head())


# ============================================================
# 8. Verify Min-Max Scaling
# ============================================================

print("Minimum Values after Min-Max Scaling:")
print(sklearn_minmax.min())

print("\nMaximum Values after Min-Max Scaling:")
print(sklearn_minmax.max())


# ============================================================
# 9. Label Encoding for RainToday
# ============================================================

label_encoder = LabelEncoder()

# Handle missing values
rain_today = df['RainToday'].fillna('No')

df['RainToday_Encoded'] = label_encoder.fit_transform(
    rain_today
)

print("Label Encoding Mapping:")

mapping = dict(
    zip(
        label_encoder.classes_,
        label_encoder.transform(label_encoder.classes_)
    )
)

print(mapping)

display(
    df[['RainToday', 'RainToday_Encoded']].head(10)
)


# ============================================================
# 10. One-Hot Encoding for Location
# ============================================================

location_encoded = pd.get_dummies(
    df['Location'],
    prefix='Location',
    drop_first=True
)

print("Original Location:")
display(df[['Location']].head())

print("One-Hot Encoded Location:")
display(location_encoded.head())

print("Number of original Location categories:",
      df['Location'].nunique())

print("Number of encoded columns:",
      location_encoded.shape[1])


# ============================================================
# 11. Create Final Transformed Dataset
# ============================================================

transformed_df = sklearn_minmax.copy()

# Add Label Encoded column
transformed_df['RainToday_Encoded'] = (
    df.loc[transformed_df.index, 'RainToday_Encoded']
)

# Add One-Hot Encoded Location columns
transformed_df = pd.concat(
    [
        transformed_df,
        location_encoded.loc[transformed_df.index]
    ],
    axis=1
)

print("Final Transformed Dataset:")
display(transformed_df.head())


# ============================================================
# 12. Compare Original vs Transformed Dataset
# ============================================================

print("Original Dataset:")
display(
    df.loc[
        transformed_df.index,
        features + ['RainToday', 'Location']
    ].head()
)

print("Transformed Dataset:")
display(transformed_df.head())


# ============================================================
# 13. Final Verification of Min-Max Scaling
# ============================================================

print("Final Statistical Verification")

verification = pd.DataFrame({
    'Min': sklearn_minmax.min(),
    'Max': sklearn_minmax.max()
})

display(verification)

print("\nExpected:")
print("Minimum = 0")
print("Maximum = 1")
```
